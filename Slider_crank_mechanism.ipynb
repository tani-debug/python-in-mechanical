```python
import numpy as np
import matplotlib.pyplot as plt
from numpy import pi
import sympy as sp

r = 125          # crank radius (mm)
c = 250          # connecting rod length (mm)

rpm = 500
omega = 2*pi*rpm/60       # rad/s

# SymPy variable
t = sp.symbols('t')

#symbolic displace
d = r*sp.cos(omega*t) + sp.sqrt(c**2 - r**2*(sp.sin(omega*t)*sp.sin(omega*t)))

# Derivatives with respect to TIME
v = sp.diff(d, t)
a = sp.diff(v, t)

# Convert symbolic expressions to NumPy functions
disp_fn = sp.lambdify(t, d, 'numpy')
vel_fn = sp.lambdify(t, v, 'numpy')
acc_fn = sp.lambdify(t, a, 'numpy')


def disp(t):
    return disp_fn(t)


def vel(t):
    return vel_fn(t)


def acc(t):
    return acc_fn(t)

time = np.linspace(0, 0.5, 1000)

theta = omega*time*180/pi


displacement = disp(time)
velocity = vel(time)
acceleration = acc(time)

plt.figure(figsize=(7.5, 15))
plt.subplot(3, 1, 1)
plt.plot(theta, displacement, 'r')
plt.ylabel("Displacement in mm")
plt.xlim(0, 360*3)
plt.tight_layout(pad = 3.0)
plt.grid()

plt.subplot(3, 1, 2)
plt.plot(theta, velocity, 'g')
plt.ylabel("Velocity")
plt.xlim(0, 360*3)
plt.tight_layout(pad = 3.0)
plt.grid()

plt.subplot(3, 1, 3)
plt.plot(theta, acceleration, 'b')
plt.ylabel("Acceleration")
plt.xlabel('Angle of Crank in degree')
plt.xlim(0, 360*3)


plt.xlim(0, 360*3)

plt.grid()

plt.show()
```

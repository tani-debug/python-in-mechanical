```python
import numpy as np
import matplotlib.pyplot as plt

#inputs
L = 10 #m
w = 50 # N/m

#Reaction
R = w*L/2
x = np.linspace(0, L, 100)

X = []
SF = []
M = []
for l in x:
    sf = R - (w*l)  #general equation of shear force
    m = (R*l) - (w*l**2/2) #general equation of bending moment
    X.append(l)
    SF.append(sf)
    M.append(m)

plt.figure(figsize=(7.5, 5), dpi=200)
plt.subplot(2, 1, 1)
plt.plot(X, SF, color='green')
plt.fill_between(X, SF, color="green", alpha=0.5, hatch="||")
plt.title("Shear Force Diagram")
plt.ylabel("Shear Force")
plt.tight_layout(pad=4.0)

plt.subplot(2, 1, 2)
plt.plot(X, M, 'red')
plt.fill_between(X, M, color="red", alpha=0.5, hatch="||")
plt.title("Bending Moment Diagram")
plt.ylabel("Bending Moment") 
plt.tight_layout(pad=4.0)

plt.show()
    
  
```

<span style="font-size:20pt; font-weight:bold">**Shear Force and Bending Moment Diagram under UDL at a particular Length**</span>

```python
import numpy as np
import matplotlib.pyplot as plt

w = 50 # udl N/m
L = 10 # Length of the beam

a = 2.5
b = 5
c = L-(a+b)

R1 = (w*b/L)*(c + b/2)
R2 = (w*b/L)*(a + b/2)
l = np.linspace(0, L, 200)

X = []
SF = []
M= []

for x in l:
    if x<a:
        sf = R1
        m = R1*x
    elif a<x<(a+b):
        sf = R1 - w*(x-a)
        m = R1*x - (w*(x-a)**2)/2
    elif x>(a+b) :
        sf = -R2
        m = R2*(L-x)
    X.append(x)
    SF.append(sf)
    M.append(m)

#for shear force diagram
plt.figure(figsize=(10, 10), dpi=200)
plt.subplot(2, 1, 1)
plt.plot(X, SF, 'green')
plt.fill_between(X, SF, color="green", hatch="||", alpha=0.5)
plt.title("Shear Force Diagram")
plt.xlabel("Length of Beam in m")
plt.ylabel("Shear Force in N")

plt.tight_layout(pad=3.0) #to keep space between two plots
#for bending moment diagram
plt.subplot(2, 1, 2)
plt.plot(X, M, 'r')
plt.fill_between(X, M, color='red', hatch="\\", alpha=0.5)
plt.title("Bending Moment Diagram")
plt.xlabel("length of Beam in m")
plt.ylabel("Moment in Nm")

plt.show()
```
